# Validation Walkthrough

This notebook is an optional walkthrough artifact for the validation package.

It reads the latest validation outputs from:
- `tests/test-results.json`
- `tests/test-results.md`

```python
import json
from pathlib import Path

repo_root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
results_path = repo_root / 'tests' / 'test-results.json'

with results_path.open('r', encoding='utf-8') as f:
    results = json.load(f)

results['summary']
```

```python
for check in results['checks']:
    status = 'PASS' if check['passed'] else 'FAIL'
    print(f"{check['id']}: {status} - {check['name']}")
    if check['details']:
        print('  details:', '; '.join(check['details']))
```

```python
markdown_path = repo_root / 'tests' / 'test-results.md'
print(markdown_path.read_text(encoding='utf-8'))
```
